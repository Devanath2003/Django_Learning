```python
nums = [1,2,2,3,3,3]

freq_dict = {}

for i in nums:
    
    freq_dict[i] = 1 + freq_dict.get(i,0)


print(freq_dict)
```

```python
max = 0
for key,val in freq_dict.items():
    if val > max:
        max = val
        freq = key

print(freq,val)
```

```python
sent = "Python is good python is easy python"
sent = sent.lower()
words = sent.split()
freq_word = {}

for w in words:
    freq_word[w] = 1 + freq_word.get(w,0)

print(freq_word)

print(max(freq_word,key=freq_word.get))

```

max(dict) = maximum key
max(dict.values()) = maximum value
max(dict, key=dict.get) = key with maximum value

```python
dictd = {}

for i in range(1,6):
    dictd[i] = i + i

print(dictd)

squares = {
    i : i * i
    for i in range(1,6)
}

print(squares)
```

```python
squares = {
    i: i * i
    for i in range(1,11)
    if i%2 == 0 and i!=6
}

print(squares)
```

```python
reverse_dict = {
    value: key
    for key,value in squares.items()
}

print(reverse_dict)
```
